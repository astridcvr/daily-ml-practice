```python
# ============================================================
# Day 12: Encoding / Jour 12 : Encodage
# Apply one-hot encoding to categorical variables.
# Appliquer l'encodage one-hot aux variables catégorielles.
# ============================================================

# === 1. Import libraries / Importer les bibliothèques ===
import pandas as pd

# === 2. Load dataset / Charger le jeu de données ===
url = "https://raw.githubusercontent.com/astridcvr/daily-ml-practice/main/data/cleaned_sales_data.csv"
df = pd.read_csv(url, encoding="latin1")

print(f"Dataset loaded: {df.shape[0]} rows, {df.shape[1]} columns")

# === 3. Identify categorical columns / Identifier les colonnes catégorielles ===
categorical_cols = df.select_dtypes(include=['object']).columns.tolist()

print("\nCategorical columns detected / Colonnes catégorielles détectées :")
print(categorical_cols)

# === 4. Select a subset for encoding / Sélectionner un sous-ensemble pour l'encodage ===
# (To avoid generating too many dummy columns)
cols_to_encode = ['STATUS', 'PRODUCTLINE', 'COUNTRY']

# === 5. Apply one-hot encoding / Appliquer l'encodage one-hot ===
df_encoded = pd.get_dummies(df, columns=cols_to_encode, drop_first=True)

print(f"\n✅ One-hot encoding applied! / Encodage one-hot appliqué !")
print(f"New shape: {df_encoded.shape}")

# === 6. Save the encoded dataset / Enregistrer le jeu de données encodé ===
output_path = "sales_data_encoded.csv"
df_encoded.to_csv(output_path, index=False)

print(f"\n💾 File saved as '{output_path}'")
print("✅ Encoding complete! / Encodage terminé !")

# === 7. Display sample output / Afficher un échantillon de sortie ===
print("\nSample of encoded columns / Exemple de colonnes encodées :")
encoded_cols = [col for col in df_encoded.columns if any(prefix in col for prefix in cols_to_encode)]
print(df_encoded[encoded_cols].head())
```
