```python
# ============================================
# Day 25 — Load Model with joblib.load()
# Load model from local upload
# ============================================

from google.colab import files
import joblib
import pandas as pd

# --- Upload model file ---
print("Select the model file: day_24_random_forest_model.joblib")
uploaded = files.upload()

# Get uploaded filename
model_filename = list(uploaded.keys())[0]
print("File uploaded:", model_filename)

# --- Load model ---
print("\nLoading model...")
model = joblib.load(model_filename)
print("Model loaded successfully.")

# ============================================
# Load test dataset from GitHub
# ============================================

url_X_test = "https://raw.githubusercontent.com/astridcvr/daily-ml-practice/main/data/X_test.csv"
url_y_test = "https://raw.githubusercontent.com/astridcvr/daily-ml-practice/main/data/y_test.csv"

print("\nLoading test dataset from GitHub...")

X_test = pd.read_csv(url_X_test)
y_test = pd.read_csv(url_y_test)

print("\nShapes:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)

# Select numeric columns only
numeric_cols = X_test.select_dtypes(include=["int64", "float64"]).columns
X_test_numeric = X_test[numeric_cols]

print("\nNumeric test dataset:", X_test_numeric.shape)

# ============================================
# Make predictions
# ============================================

print("\nRunning predictions...")
y_pred = model.predict(X_test_numeric)

print("Predictions completed. First 5 results:")
print(y_pred[:5])
```
