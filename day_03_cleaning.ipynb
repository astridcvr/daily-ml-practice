```python
# ============================================================
# Day 03: Data Cleaning / Jour 3 : Nettoyage des données
# Handle missing values using mean/mode imputation and remove
# columns with excessive nulls.
# Gérer les valeurs manquantes avec l’imputation (moyenne/modale)
# et supprimer les colonnes trop incomplètes.
# ============================================================

import pandas as pd

# === 1. Load dataset from GitHub / Charger le jeu de données depuis GitHub ===
url = "https://raw.githubusercontent.com/astridcvr/daily-ml-practice/main/data/sales_data_sample.csv"
df = pd.read_csv(url, encoding="latin1")

print(f"Dataset shape: {df.shape}")
print(f"Columns: {list(df.columns)}")

# === 2. Detect missing values / Détecter les valeurs manquantes ===
missing = df.isnull().sum()
print("\nColumns with missing values / Colonnes avec valeurs manquantes :")
print(missing[missing > 0])

missing_summary = pd.DataFrame({
    "Column": missing[missing > 0].index,
    "MissingValues": missing[missing > 0].values
})
print("\nSummary of missing values / Résumé des valeurs manquantes :")
print(missing_summary)

# === 3. Clean data / Nettoyage des données ===
df_cleaned = df.copy()

# 3.1 Remove columns with >50% missing / Supprimer les colonnes >50% valeurs manquantes
threshold = len(df_cleaned) * 0.5
df_cleaned = df_cleaned.dropna(axis=1, thresh=threshold)

# 3.2 Identify numeric and categorical columns / Identifier colonnes numériques et catégorielles
num_cols = df_cleaned.select_dtypes(include=["float64", "int64"]).columns
cat_cols = df_cleaned.select_dtypes(include=["object"]).columns

# 3.3 Fill numeric columns with mean / Remplir numériques avec la moyenne
for col in num_cols:
    df_cleaned[col] = df_cleaned[col].fillna(df_cleaned[col].mean())

# 3.4 Fill categorical columns with mode / Remplir catégorielles avec mode
for col in cat_cols:
    if df_cleaned[col].isnull().any():
        df_cleaned[col] = df_cleaned[col].fillna(df_cleaned[col].mode()[0])

# === 4. Verify cleaning / Vérifier nettoyage ===
print("\nMissing values after cleaning / Valeurs manquantes après nettoyage :")
print(df_cleaned.isnull().sum().sum())

print("\n✅ Data cleaning complete! / Nettoyage des données terminé !")

# === 5. Save cleaned CSV / Sauvegarder le CSV nettoyé ===
df_cleaned.to_csv("cleaned_sales_data.csv", index=False)
print("\n💾 Cleaned file saved as 'cleaned_sales_data.csv'")
```
