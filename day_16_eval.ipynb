```python
# ============================================================
# Day 16: Model Evaluation / Jour 16 : Évaluation du modèle
# Compute MSE and R² after loading the trained regression model.
# Calculer le MSE et le R² après avoir chargé le modèle entraîné.
# ============================================================

# === 1. Import libraries / Importer les bibliothèques ===
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# === 2. Load datasets / Charger les jeux de données ===
X_train = pd.read_csv("https://raw.githubusercontent.com/astridcvr/daily-ml-practice/main/data/X_train.csv")
X_test  = pd.read_csv("https://raw.githubusercontent.com/astridcvr/daily-ml-practice/main/data/X_test.csv")
y_train = pd.read_csv("https://raw.githubusercontent.com/astridcvr/daily-ml-practice/main/data/y_train.csv")
y_test  = pd.read_csv("https://raw.githubusercontent.com/astridcvr/daily-ml-practice/main/data/y_test.csv")

print("Datasets loaded:")
print("X_train:", X_train.shape, "X_test:", X_test.shape)
print("y_train:", y_train.shape, "y_test:", y_test.shape)

# === 3. Keep only numeric columns / Garder uniquement les colonnes numériques ===
X_train_num = X_train.select_dtypes(include="number")
X_test_num  = X_test.select_dtypes(include="number")

print("\nNumeric datasets only:")
print("X_train:", X_train_num.shape, "X_test:", X_test_num.shape)

# === 4. Initialize and train model / Initialiser et entraîner le modèle ===
model = LinearRegression()
model.fit(X_train_num, y_train)

print("\nModel loaded and trained / Modèle chargé et entraîné")

# === 5. Make predictions / Faire des prédictions ===
y_pred = model.predict(X_test_num)

# === 6. Compute metrics / Calculer les métriques ===
mse = mean_squared_error(y_test, y_pred)
r2  = r2_score(y_test, y_pred)

print("\nEvaluation Metrics / Métriques d'évaluation:")
print("MSE:", mse)
print("R² :", r2)

# === 7. Save evaluation results / Enregistrer les résultats ===
output_path = "model_eval_results.csv"  # Save locally in Colab

eval_df = pd.DataFrame({
    "Metric": ["MSE", "R2"],
    "Value": [mse, r2]
})

eval_df.to_csv(output_path, index=False)

print("\nEvaluation results saved locally / Résultats enregistrés localement :")
eval_df
```
