```python
# ============================================================
# Day 13: Scaling / Jour 13 : Normalisation
# Apply StandardScaler to numerical features for model training.
# Appliquer StandardScaler aux variables numériques pour
# standardiser les données avant l'entraînement des modèles.
# ============================================================

# === 1. Import libraries / Importer les bibliothèques ===
import pandas as pd
from sklearn.preprocessing import StandardScaler

# === 2. Load dataset / Charger le jeu de données ===
url = "https://raw.githubusercontent.com/astridcvr/daily-ml-practice/main/data/cleaned_sales_data.csv"
df = pd.read_csv(url, encoding="latin")

print("Dataset loaded / Jeu de données chargé")
df.head()

# === 3. Select numerical columns / Sélectionner les colonnes numériques ===
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()

print("Numerical columns selected / Colonnes numériques sélectionnées :")
print(numerical_cols)

# === 4. Initialize the scaler / Initialiser le scaler ===
scaler = StandardScaler()

# === 5. Fit and transform / Ajuster et transformer ===
df_scaled = df.copy()
df_scaled[numerical_cols] = scaler.fit_transform(df[numerical_cols])

print("Scaling complete / Normalisation terminée")

# === 6. Create an output preview / Créer un aperçu du résultat ===
print("\nPreview of scaled data / Aperçu des données normalisées :")
print(df_scaled.head())

# === 7. Export scaled dataset (for manual download) / Exporter les données ===
output_filename = "scaled_sales_data.csv"
df_scaled.to_csv(output_filename, index=False)

output_filename
```
