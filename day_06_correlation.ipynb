```python
# ===============================================================
# Day 06 – Correlation / Corrélation
# ===============================================================

"""
🎯 Objective / Objectif :
EN: Explore correlations between numeric features to identify relationships.
FR: Explorer les corrélations entre les variables numériques pour identifier les relations.

📘 Description :
EN: We will calculate a correlation matrix and visualize it using a heatmap.
FR: Nous calculerons une matrice de corrélation et la visualiserons avec une carte thermique.

📅 Author / Auteur : Astrid Villalobos
📍 Montréal, QC
"""

# ===============================================================
# === 1. Imports ===
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ===============================================================
# === 2. Load dataset / Charger le jeu de données ===
url = "https://raw.githubusercontent.com/astridcvr/daily-ml-practice/main/data/sales_data_features.csv"
df = pd.read_csv(url, encoding="latin1")
print(f"Dataset loaded: {df.shape[0]} rows, {df.shape[1]} columns")
print(f"Columns: {list(df.columns)}")

# ===============================================================
# === 3. Select numeric features / Sélectionner les variables numériques ===
num_cols = [
    "SALES",
    "QUANTITYORDERED",
    "PRICEEACH",
    "TOTAL_ORDER_SALES",
    "AVG_PRICE_ORDER",
    "SALES_RATIO"
]

df_numeric = df[num_cols]

# ===============================================================
# === 4. Compute correlation matrix / Calculer la matrice de corrélation ===
corr_matrix = df_numeric.corr()
print("\nCorrelation matrix / Matrice de corrélation :")
print(corr_matrix)

# ===============================================================
# === 5. Visualize correlation heatmap / Visualiser la carte thermique ===
plt.figure(figsize=(10, 8))
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)
plt.title("EN: Correlation Heatmap | FR: Carte thermique de corrélation")
plt.show()

```

SALES vs QUANTITYORDERED → likely strong positive correlation

SALES vs PRICEEACH → likely moderate positive correlation

TOTAL_ORDER_SALES vs SALES → likely very strong positive correlation

SALES_RATIO → ratios with other columns will vary
