```python
# ===============================================================
# Day 09 – Pivot Tables / Jour 09 : Tableaux croisés
# ===============================================================

"""
🎯 Objective / Objectif :
EN: Create pivot tables to analyze sales by country and product line.
FR: Créer des tableaux croisés pour analyser les ventes par pays et par ligne de produit.

📘 Description :
EN: Pivot tables help summarize data to reveal relationships between multiple variables.
FR: Les tableaux croisés permettent de résumer les données et de révéler les relations entre plusieurs variables.

📅 Author / Auteur : Astrid Villalobos
📍 Montréal, QC
"""

# ===============================================================

# === 1. Imports / Importations ===
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# === 2. Load dataset / Charger le jeu de données ===
url = "https://raw.githubusercontent.com/astridcvr/daily-ml-practice/main/data/sales_data_features.csv"
df = pd.read_csv(url, encoding="latin1")

print(f"✅ Dataset loaded: {df.shape[0]} rows, {df.shape[1]} columns")

# === 3. Create pivot table / Créer le tableau croisé ===
pivot_table = pd.pivot_table(
    df,
    values="SALES",
    index="COUNTRY",
    columns="PRODUCTLINE",
    aggfunc="sum",
    fill_value=0
)

print("\n📊 Pivot Table (Country vs Product Line) / Tableau croisé (Pays vs Ligne de produit) :")
print(pivot_table.head())

# === 4. Save pivot table to CSV / Sauvegarder le tableau croisé ===
pivot_table.to_csv("country_vs_product_pivot.csv")
print("\n💾 File saved as 'country_vs_product_pivot.csv'")

# === 5. Visualization / Visualisation ===
plt.figure(figsize=(10, 6))
sns.heatmap(pivot_table, cmap="Blues", annot=False, linewidths=0.5)
plt.title("EN: Total Sales by Country and Product Line\nFR: Ventes totales par pays et ligne de produit")
plt.xlabel("Product Line / Ligne de produit")
plt.ylabel("Country / Pays")
plt.tight_layout()
plt.show()

print("\n✅ Pivot table analysis complete! / Analyse du tableau croisé terminée !")
```
