```python
# ============================================================
# Day 11: Time Series / Jour 11 : Séries temporelles
# Analyze monthly sales trend over time.
# Analyser la tendance des ventes mensuelles dans le temps.
# ============================================================

# === 1. Import libraries / Importer les bibliothèques ===
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# === 2. Load monthly sales data / Charger les ventes mensuelles ===
url = "https://raw.githubusercontent.com/astridcvr/daily-ml-practice/main/data/sales_by_month.csv"
df = pd.read_csv(url)

print(f"✅ Dataset loaded: {df.shape[0]} rows, {df.shape[1]} columns")
print("📊 Columns:", list(df.columns))

# === 3. Convert 'PERIOD' to datetime / Convertir 'PERIOD' en datetime ===
df['PERIOD'] = pd.to_datetime(df['PERIOD'], format='%Y-%m', errors='coerce')

# === 4. Sort data by date / Trier les données par date ===
df = df.sort_values('PERIOD')

# === 5. Plot time series / Tracer la série temporelle ===
plt.figure(figsize=(10, 5))
sns.lineplot(data=df, x='PERIOD', y='SALES', marker='o', linewidth=2)
plt.title("Monthly Sales Trend / Tendance des ventes mensuelles")
plt.xlabel("Date / Date")
plt.ylabel("Sales / Ventes")
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

# === 6. Compute basic trend metrics / Calculer des métriques de tendance ===
mean_sales = df['SALES'].mean()
max_sales = df['SALES'].max()
min_sales = df['SALES'].min()
growth_rate = ((df['SALES'].iloc[-1] - df['SALES'].iloc[0]) / df['SALES'].iloc[0]) * 100

# === 7. Display insights / Afficher les observations ===
print("\n📈 Time Series Insights / Observations de la série temporelle :")
print(f"Average monthly sales / Moyenne mensuelle : {mean_sales:,.2f}")
print(f"Highest month sales / Mois avec ventes max : {max_sales:,.2f}")
print(f"Lowest month sales / Mois avec ventes min : {min_sales:,.2f}")
print(f"Overall growth rate / Taux de croissance global : {growth_rate:.2f}%")

# === 8. Save analyzed data / Sauvegarder les données analysées ===
df.to_csv("sales_monthly_trend.csv", index=False)
print("\n💾 File saved as 'sales_monthly_trend.csv'")
```
