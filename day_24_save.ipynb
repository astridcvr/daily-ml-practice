```python
# ============================================
# Day 24 – Save Model / Sauvegarde
# Model persistence with joblib
# ============================================

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
import joblib

# ============================================
# 1. Load datasets from GitHub
# ============================================

print("Loading datasets from GitHub...\n")

X = pd.read_csv("https://raw.githubusercontent.com/astridcvr/daily-ml-practice/main/data/X_train.csv")
y = pd.read_csv("https://raw.githubusercontent.com/astridcvr/daily-ml-practice/main/data/y_train.csv")

# y must be 1D
y = y.values.ravel()

print("Shapes:")
print("X:", X.shape)
print("y:", y.shape)

# ============================================
# 2. Keep only numeric columns
# ============================================

numeric_cols = X.select_dtypes(include=['int64', 'float64']).columns.tolist()
X_num = X[numeric_cols]

print("\nNumeric dataset:", X_num.shape)

# ============================================
# 3. Train/test split
# ============================================

X_train, X_test, y_train, y_test = train_test_split(
    X_num, y, test_size=0.2, random_state=42
)

# ============================================
# 4. Train model (Random Forest)
# ============================================

print("\nTraining Random Forest model...")

model = RandomForestRegressor(
    n_estimators=300,
    random_state=42
)

model.fit(X_train, y_train)

print("Model trained successfully.")

# ============================================
# 5. Evaluate the model
# ============================================

y_pred = model.predict(X_test)

mse = mean_squared_error(y_test, y_pred)
r2  = r2_score(y_test, y_pred)

print("\nEvaluation Metrics:")
print("MSE:", mse)
print("R² :", r2)

# ============================================
# 6. Save the model using joblib
# ============================================

model_path = "day_24_random_forest_model.joblib"

joblib.dump(model, model_path)

print(f"\nModel saved successfully as: {model_path}")

# ============================================
# 7. Save evaluation metrics
# ============================================

eval_df = pd.DataFrame({
    "Metric": ["MSE", "R2"],
    "Value": [mse, r2]
})

output_file = "day_24_save_model_results.csv"
eval_df.to_csv(output_file, index=False)

print(f"Evaluation metrics saved: {output_file}")
eval_df
```
