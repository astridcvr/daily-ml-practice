```python
# ============================================================
# Day 19: Random Forest / Jour 19 : Forêt aléatoire
# Train a robust Random Forest model and evaluate performance.
# Entraîner un modèle Random Forest robuste et évaluer ses performances.
# ============================================================

# === 1. Import libraries ===
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score

# === 2. Load datasets from GitHub ===
print("Loading datasets from GitHub...\n")

X_train = pd.read_csv("https://raw.githubusercontent.com/astridcvr/daily-ml-practice/main/data/X_train.csv")
X_test  = pd.read_csv("https://raw.githubusercontent.com/astridcvr/daily-ml-practice/main/data/X_test.csv")
y_train = pd.read_csv("https://raw.githubusercontent.com/astridcvr/daily-ml-practice/main/data/y_train.csv")
y_test  = pd.read_csv("https://raw.githubusercontent.com/astridcvr/daily-ml-practice/main/data/y_test.csv")

print("Shapes:")
print("X_train:", X_train.shape, "X_test:", X_test.shape)
print("y_train:", y_train.shape, "y_test:", y_test.shape)

# === 3. Keep only numeric columns ===
X_train_num = X_train.select_dtypes(include=[np.number])
X_test_num  = X_test.select_dtypes(include=[np.number])

print("\nNumeric dataset:")
print("X_train_num:", X_train_num.shape, "X_test_num:", X_test_num.shape)

# === 4. Train Random Forest model ===
print("\nTraining Random Forest model...")

rf = RandomForestRegressor(
    n_estimators=200,
    max_depth=None,
    random_state=42,
    n_jobs=-1
)

rf.fit(X_train_num, y_train.values.ravel())

print("Model trained successfully.")

# === 5. Predictions ===
y_pred = rf.predict(X_test_num)

# === 6. Compute metrics ===
mse = mean_squared_error(y_test, y_pred)
r2  = r2_score(y_test, y_pred)

print("\nEvaluation Metrics / Métriques d'évaluation:")
print("MSE:", mse)
print("R² :", r2)

# === 7. Save results ===
results = pd.DataFrame({
    "Metric": ["MSE", "R2"],
    "Value": [mse, r2]
})

output_path = "day_19_rf_results.csv"
results.to_csv(output_path, index=False)

print("\nRandom Forest results saved locally:", output_path)
print(results)
```
