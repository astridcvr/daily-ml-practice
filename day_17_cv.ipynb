```python
# ============================================================
# Day 17: Cross Validation / Jour 17 : Validation croisée
# Validate model performance using K-Fold cross-validation.
# Valider les performances du modèle avec la validation croisée.
# ============================================================

# === 1. Import libraries / Importer les bibliothèques ===
import pandas as pd
import numpy as np
from sklearn.model_selection import KFold, cross_val_score
from sklearn.linear_model import LinearRegression

# === 2. Load datasets from GitHub / Charger les jeux depuis GitHub ===
print("Loading datasets from GitHub...\n")

X = pd.read_csv("https://raw.githubusercontent.com/astridcvr/daily-ml-practice/main/data/X_train.csv")
y = pd.read_csv("https://raw.githubusercontent.com/astridcvr/daily-ml-practice/main/data/y_train.csv")

print("Shapes:")
print(f"X: {X.shape}")
print(f"y: {y.shape}\n")

# === 3. Keep only numeric data / Conserver uniquement les données numériques ===
X_numeric = X.select_dtypes(include=['int64', 'float64'])

print("Numeric dataset:")
print(f"X_numeric: {X_numeric.shape}\n")

# === 4. Initialize model / Initialiser le modèle ===
model = LinearRegression()

# === 5. Set up K-Fold / Définir la validation croisée K-Fold ===
kf = KFold(n_splits=5, shuffle=True, random_state=42)

# === 6. Cross-validation scores / Calculer les scores CV ===
mse_scores = -cross_val_score(model, X_numeric, y, cv=kf, scoring='neg_mean_squared_error')
r2_scores = cross_val_score(model, X_numeric, y, cv=kf, scoring='r2')

print("Cross-validation results:")
print("MSE scores:", mse_scores)
print("R² scores :", r2_scores)

print("\nAverage Metrics:")
print(f"Average MSE: {mse_scores.mean()}")
print(f"Average R² : {r2_scores.mean()}")

# === 7. Create results dataframe / Créer un tableau de résultats ===
cv_results = pd.DataFrame({
    "Metric": ["MSE_mean", "MSE_std", "R2_mean", "R2_std"],
    "Value": [
        mse_scores.mean(),
        mse_scores.std(),
        r2_scores.mean(),
        r2_scores.std()
    ]
})

# === 8. Save results locally so you can upload to GitHub ===
output_path = "day_17_cv_results.csv"
cv_results.to_csv(output_path, index=False)

print("\nCross-validation results saved locally:", output_path)
cv_results
```
