```python
# ===============================================================
# Day 05 – Visualization / Visualisation
# ===============================================================

"""
🎯 Objective / Objectif :
EN: Visualize the top countries by total sales using bar charts.
FR: Visualiser les pays les plus performants en termes de ventes à l’aide de graphiques à barres.

📘 Description :
EN: This notebook focuses on data visualization using Matplotlib and Seaborn.
    The goal is to highlight the top-performing countries by sales volume and
    analyze overall distribution visually.
FR: Ce notebook met l’accent sur la visualisation des données avec Matplotlib et Seaborn.
    L’objectif est de mettre en évidence les pays les plus performants en termes de ventes
    et d’analyser la distribution globale visuellement.

📅 Author / Auteur : Astrid Villalobos
📍 Montréal, QC
"""

# === 1. Import Libraries / Importer les bibliothèques ===
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# === 2. Load Cleaned Dataset / Charger le jeu de données nettoyé ===
url = "https://raw.githubusercontent.com/astridcvr/daily-ml-practice/main/sales_data_sample.csv"
df = pd.read_csv(url, encoding='latin1')

# === 3. Compute Total Sales by Country / Calculer les ventes totales par pays ===
sales_by_country = df.groupby("COUNTRY")["SALES"].sum().sort_values(ascending=False).reset_index()

# === 4. Select Top 10 Countries / Sélectionner les 10 premiers pays ===
top_countries = sales_by_country.head(10)

# === 5. Plot Top Countries / Tracer les meilleurs pays ===
plt.figure(figsize=(10, 6))
sns.barplot(data=top_countries, x="SALES", y="COUNTRY", palette="viridis")
plt.title("Top 10 Countries by Total Sales / Top 10 des pays par ventes totales", fontsize=14)
plt.xlabel("EN: Total Sales | FR: Ventes Totales", fontsize=12)
plt.ylabel("EN: Country | FR: Pays", fontsize=12)
plt.tight_layout()
plt.show()

# === 6. Add Pie Chart (Optional) / Ajouter un diagramme circulaire (optionnel) ===
plt.figure(figsize=(8, 8))
plt.pie(
    top_countries["SALES"],
    labels=top_countries["COUNTRY"],
    autopct="%1.1f%%",
    startangle=140,
    colors=sns.color_palette("viridis", len(top_countries))
)
plt.title("Sales Share by Country / Part des ventes par pays", fontsize=14)
plt.show()

# === 7. Summary / Résumé ===
"""
✅ Summary / Résumé :
EN: The visualization clearly shows which countries contribute the most to sales.
    The top 3 typically represent the majority of total revenue, highlighting key markets.
FR: La visualisation montre clairement quels pays contribuent le plus aux ventes.
    Les 3 premiers représentent généralement la majorité du chiffre d’affaires total,
    ce qui met en évidence les marchés clés.

📅 Next Step / Prochaine étape :
Day 06 – Correlation Analysis / Analyse de corrélation.
"""
```
