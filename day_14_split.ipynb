```python
# ============================================================
# Day 14: Train/Test Split / Jour 14 : Division des données
# Split the dataset into training and testing sets for modeling.
# Diviser le jeu de données en ensembles d'entraînement et de test.
# ============================================================

# === 1. Import libraries / Importer les bibliothèques ===
import pandas as pd
from sklearn.model_selection import train_test_split

# === 2. Load encoded data / Charger les données encodées ===
url = "https://raw.githubusercontent.com/astridcvr/daily-ml-practice/main/data/sales_data_encoded.csv"
df = pd.read_csv(url)

print("Dataset loaded. Shape / Jeu de données chargé. Dimensions :", df.shape)

# === 3. Define features and target / Définir les variables explicatives et la cible ===
X = df.drop(columns=['SALES'])   # Features / Variables explicatives
y = df['SALES']                  # Target / Variable cible

# === 4. Perform train/test split / Effectuer la division train/test ===
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,       # 20% test / 20% test
    random_state=42,     # Reproducibility / Reproductibilité
)

# === 5. Show results / Afficher les résultats ===
print("\nShapes after split / Dimensions après division :")
print("X_train :", X_train.shape)
print("X_test  :", X_test.shape)
print("y_train :", y_train.shape)
print("y_test  :", y_test.shape)

# === 6. Save split datasets / Enregistrer les ensembles séparés ===
X_train.to_csv("X_train.csv", index=False)
X_test.to_csv("X_test.csv", index=False)
y_train.to_csv("y_train.csv", index=False)
y_test.to_csv("y_test.csv", index=False)

print("\nCSV files generated. Download and upload to your GitHub repo.")
print("Fichiers CSV générés. Télécharge-les et ajoute-les à ton dépôt GitHub.")
```
